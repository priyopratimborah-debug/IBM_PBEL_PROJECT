```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D,MaxPooling2D,Flatten,Dense
```

```python
(x_train,y_train),(x_test,y_test) = tf.keras.datasets.fashion_mnist.load_data()
```

```python
plt.imshow(x_train[14],cmap = 'gray')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(10,6))
for i in range(12):
    plt.subplot(3,4,i+1)
    plt.imshow(x_train[i],cmap = 'gray')
    plt.title(f'Label: {y_train[i]}')
    plt.axis('off')
```

```python
class_names = [
    'T-shirt/top',
    'Trouser',
    'Pullover',
    'Dress',
    'Coat',
    'Sandal',
    'Shirt',
    'Sneaker',
    'Bag',
    'Ankle boot'
]
```

```python
x_train = x_train.reshape(-1,28,28,1)
x_test = x_test.reshape(-1,28,28,1)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
.0/255.0
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
model = Sequential()
```

```python
model.add(Conv2D(32,(3,3),activation ='relu',input_shape=(28,28,1)))
```

```python
model.add(MaxPooling2D(pool_size=(2,2)))
```

```python
model.add(Flatten())
```

```python
model.add(Dense(128,activation='relu'))
```

```python
model.add(Dense(10,activation= 'softmax'))
```

```python
model.summary()
```

```python
model.compile(
    optimizer = 'adam',
    loss = 'sparse_categorical_crossentropy',
    metrics = ['accuracy']
)
```

```python
history = model.fit(
    x_train,y_train,
    epochs = 5,
    batch_size = 64,
    validation_split = 0.2
)
```

```python
test_loss,test_accuracy = model.evaluate(
    x_test,y_test
)
```

```python
print(f"accuracy = {test_accuracy}, loss ={test_loss}")
```

```python
plt.figure(figsize=(6,4))
plt.plot(
    history.history['accuracy'],label = 'Training accuracy '
)
plt.plot(
    history.history['val_accuracy'],label = 'Validation accuracy '
)
plt.xlabel("Epochs")
plt.ylabel("accuracy")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(
    history.history['loss'],label = 'Training Loss'
)
plt.plot(
    history.history['val_loss'],label = 'Validation Loss'
)
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()

plt.show()
```
